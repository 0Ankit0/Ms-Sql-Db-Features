# Project Structure and Update Guide

## Repository intent
This project is a practical notebook collection for SQL Server and ASP.NET development notes.

## Folder and file structure
- Notebook files (`*.ipynb`) hold topic-focused examples and snippets.
- `_toc.yml` defines book ordering and grouping for navigation.
- `_config.yml` contains Jupyter Book configuration metadata.
- `README.md` gives a high-level overview and usage guidance.


## How to keep notebooks consistent

1. Start every notebook with a short **Overview** markdown section.
2. Group examples by use case and add headings before major code blocks.
3. End with **Notes** or **Gotchas** when behavior is environment-dependent.
4. Update `_toc.yml` whenever new notebooks are added.


## Suggested categories

- Core SQL patterns
- SQL Server features
- Data conversion and JSON workflows
- ASP.NET utilities and frontend integration
- Testing and development tooling
